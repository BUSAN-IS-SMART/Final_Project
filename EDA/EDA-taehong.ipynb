### 부두별 작업 효율을 보려면?

$$
\text{부두가동률(\%)} =
\frac{\text{당월 실제 처리실적(TEU)}}{\text{연간 설계하역능력(TEU)} / 12}
\times 100
$$

-> 구체적으로 날짜를 대입할 때는 다음과 같이

$$
\text{월간 환산 설계능력(TEU)} =
\text{연간 설계하역능력(TEU)}
\times
\frac{\text{해당 월 일수}}{\text{해당 연도 총 일수}}
$$

ex) 신항 1부두 2024년 2월의 환산은?

$$
209만(TEU)
\times
\frac{\text{29}}{\text{366}}
$$


도출된 부두가동률과 부두별 월 재항시간 평균값을 비교한다.

결론적으로 물동량을 계산한 값(**부두가동률**)과 시간을 계산한 값(**재항시간 평균**)을 비교했을 때,
**운영 효율**을 정의할 수 있다.


| 월 | 부두가동률(%) | 재항시간 평균 |
|:---:|:---:|:---:|
| 1월 | 110% | 16시간 |
| 2월 | 120% | 17시간 |
| 3월 | 140% | 22시간 |
| 4월 | 135% | 18시간 |
| 5월 | 130% | 17시간 |


-> 3월에 140%를 임계점으로 파악


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

def load_dataset(filename):
    BASE_DIR = './data'
    EXTENSION = '.csv'
    TARGET_PATH = f'{BASE_DIR}/{filename}{EXTENSION}'
    try:
        return pd.read_csv(TARGET_PATH)
    except FileNotFoundError:
        return None
    except UnicodeDecodeError:
        return pd.read_csv(TARGET_PATH, encoding = 'euc-kr')
```

```python
import pandas as pd
from pathlib import Path

# 1. 파일 경로 설정
base = Path.cwd()

# PyCharm 실행 위치가 프로젝트 루트 또는 EDA 폴더인 경우 모두 대응
file_path = base / "data" / "events.xlsx"

if not file_path.exists():
    file_path = base.parent / "data" / "events.xlsx"

print("파일 경로:", file_path)
print("파일 존재 여부:", file_path.exists())

# 2. Excel 사건 목록 불러오기
df = pd.read_excel(
    file_path,
    sheet_name="사건목록_30건",
    header=3
)

# 3. 날짜 형식 변환
df["게시일"] = pd.to_datetime(df["게시일"], errors="coerce")

# 4. 전체 사건 목록 조회
display(df[[
    "부두",
    "게시일",
    "사건·공식 공지 내용",
    "공지상 운영 영향"
]])

# 5. 2024년 2월 2부두 사건 조회
result = df[
    (df["부두"] == "2부두") &
    (df["게시일"].dt.strftime("%Y-%m") == "2024-02")
]

print("\n2024년 2월 2부두 사건")
display(result[[
    "게시일",
    "사건·공식 공지 내용",
    "공지상 운영 영향",
    "출처 URL"
]])
```
