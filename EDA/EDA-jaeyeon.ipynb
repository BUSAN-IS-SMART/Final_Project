```python
import pandas as pd
from pathlib import Path

f = Path("../data/record.csv")
df = pd.read_csv(f, encoding="cp949").dropna(how="all")

df["입항일시"] = pd.to_datetime(df["입항일시"], errors="raise")
df["출항일시"] = pd.to_datetime(df["출항일시"], errors="raise")

df["재항시간"] = (df["출항일시"] - df["입항일시"]).dt.total_seconds() / 3600

delete_data = df["재항시간"] >= 168
result = df.loc[~delete_data].copy()

save = f.with_name("record_except_168.csv")
result.to_csv(save, index=False)
```
